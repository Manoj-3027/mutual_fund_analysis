```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=aum_df,
    x='fund_house',
    y='aum_crore'
)

plt.xticks(rotation=45)

plt.title("AUM by Fund House")

plt.xlabel("Fund House")

plt.ylabel("AUM (Crores)")

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    sip_df['month'],
    sip_df['sip_inflow_crore']
)

plt.title("Monthly SIP Inflows Trend")

plt.xlabel("Month")

plt.ylabel("SIP Inflow (Crores)")

plt.grid(True)

plt.show()
```
